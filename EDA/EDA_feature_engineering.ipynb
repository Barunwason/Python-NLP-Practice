```python
import pandas as pd
import numpy as np
df=pd.read_excel('Data_Train.xlsx')
df
```

```python
df.describe()
```

```python
a=df['Date_of_Journey'].str.split('/')
df['date']=a.str[0]
df['month']=a.str[1]
df['year']=a.str[2]
df['date']=df['date'].astype(int)
df['month']=df['month'].astype(int)
df['year']=df['year'].astype(int)
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0, '2 stops':2, '1 stop':1, '3 stops':3, np.nan:1, '4 stops':4})
```

```python
a=df.size
a
```

```python
from sklearn.preprocessing import LabelEncoder
import numpy as np
encoder= LabelEncoder()
encoded=encoder.fit_transform(df[['Source']])
```

```python
df['source_encoded']=pd.DataFrame(encoded)
df['source_encoded'].unique()
```

```python
encoded2=encoder.fit_transform(df[['Destination']])
df['destination_encoded']=pd.DataFrame(encoded2)
df['destination_encoded'].unique()
df['Destination'].unique()
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
```

```python
import pandas as pd
df=pd.DataFrame({
    'color':['red','blue','green','white','black']
})
df
```

```python
encoded=encoder.fit_transform(df[['color']]).toarray()
encoded
```

```python
df_encoded=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
df_encoded
```

```python
a=encoder.transform([['red']]).toarray()
df1=pd.DataFrame(a,columns=encoder.get_feature_names_out())
df1
```

```python
pd.concat([df,df_encoded],axis=1)
```

```python
import seaborn as sns
data=sns.load_dataset('tips')
data
```

```python
b=encoder.fit_transform(data[['day']]).toarray()
df2=pd.DataFrame(b,columns=encoder.get_feature_names_out())
df2
```

```python
pd.concat([data['day'],df2],axis=1).value_counts()
```

```python
c=encoder.fit_transform(data[['sex']]).toarray()
df4=pd.DataFrame(c,columns=encoder.get_feature_names_out())
df4
```

```python
pd.concat([data['sex'],df4],axis=1)
```

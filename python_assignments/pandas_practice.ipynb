```python
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
print(df.head())
df.isnull().sum()
```

```python
df['updated_age']=df['age'].fillna(df['age'].mean())
```

```python
df.isnull().sum()
```

```python
df[['age','updated_age','deck']]
```

```python
mode_value=df[df['deck'].notna()]['deck'].mode()[0]
mode_value
```

```python
df['deck_updated']=df['deck'].fillna(mode_value)
df['deck_updated'].isnull().sum()
```

```python
df[df['deck_updated'].isnull()][['deck','deck_updated']].isnull().sum()
```

```python
df[['deck','deck_updated']]
```

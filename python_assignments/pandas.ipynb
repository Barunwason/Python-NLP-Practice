### Assignment 1: DataFrame Creation and Indexing

1. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Set the index to be the first column.
2. Create a Pandas DataFrame with columns 'A', 'B', 'C' and index 'X', 'Y', 'Z'. Fill the DataFrame with random integers and access the element at row 'Y' and column 'B'.

```python
import pandas as pd
d={
    'index':[1,2,3,4],
    'name':['barun','ram','sham','mohan'],
    'age':[21,29,32,24],
    'city':['delhi','haryana','gujrat','banglore']
}
df = pd.DataFrame(d,index=d['index'])
df
```

```python
import numpy as np 
df=pd.DataFrame(np.random.randint(1,100,size=(6,4)),columns=['A','B','C','D'])
df.set_index('A',inplace=True)
df
```

```python
df=pd.DataFrame(np.random.randint(1,100,size=(3,3)),columns=['A','B','C'],index=['X','Y','Z'])
print(df)
print(df.at['Y','B'])
```

### Assignment 3: Data Cleaning

1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.
2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.


```python
df=pd.DataFrame(np.random.randint(1,100,size=(5,3)))
df.at[1,2]=np.nan
print(df)
print(df.isnull().sum())
df[2]=df[2].fillna(df[2].mean())
df
```

```python
df=pd.DataFrame(np.random.randint(1,100,size=(5,3)))
df.at[1,2]=np.nan
print(df)
df.dropna(inplace=True) #it is used to delete the row with NaN value
print(df)
df[2]=df[2].astype(int) #converting the datatype into int of column 2
print(df)
```

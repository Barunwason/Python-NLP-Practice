### Preprocessing and Cleaning

```python
#changing the rating in 0 and 1
df['rating']=df['rating'].apply(lambda x:0 if x<3 else 1)
```

```python
df['rating'].unique()
```

```python
#checking if the data in balanced or not
df['rating'].value_counts()
```

```python
#lowering the reviewText
df['reviewText']=df['reviewText'].str.lower()
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
from bs4 import BeautifulSoup
```

```python
## Removing special characters
df['reviewText']=df['reviewText'].apply(lambda x:re.sub('[^a-z A-z 0-9-]+', '',x))
## Remove the stopswords
df['reviewText']=df['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stopwords.words('english')]))
## Remove url 
df['reviewText']=df['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove any additional spaces
df['reviewText']=df['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
## Remove html tags
df['reviewText'] = df['reviewText'].apply(lambda x: BeautifulSoup(x, 'html.parser').get_text())
```

```python
df.head()
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
df['reviewText'] = df['reviewText'].apply(lambda x:lemmatize_words(x))
```

```python
df['reviewText']
```

```python
from nltk import word_tokenize,sent_tokenize
```

```python
def tokennization(corpus):   
    words1=[]
    for sent in corpus:
        sent_token=sent_tokenize(sent)
        for sent in sent_token:
            words1.append(word_tokenize(sent))
    return words1
```

```python
words=tokennization(df['reviewText'])
```

```python
len(words)
```

```python
## Training the word2vec model
import gensim
model=gensim.models.Word2Vec(words)
```

```python
model.corpus_count
```

```python
# to get all the vocabulary
len(model.wv.index_to_key)
```

```python
model.wv.similar_by_word('library')
```

```python
model.wv['book'].shape
```

```python
import numpy as np
```

```python
def avgWord2Vec(doc):
    vectors = [model.wv[word] for word in doc if word in model.wv.index_to_key]
    if len(vectors) == 0:
        return np.zeros(model.vector_size)  # safeguard for empty sentence
    return np.mean(vectors, axis=0) 
```

```python
from tqdm import tqdm
```

```python
X=[]
for i in tqdm(range(len(words))):
    X.append(avgWord2Vec(words[i]))
```

```python
len(X)
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,df['rating'],train_size=0.75)
```

```python
len(X_test)
```

```python
#Using the Random Forest classifier ML model
from sklearn.ensemble import RandomForestClassifier

classifier=RandomForestClassifier()
```

```python
#Training the model
classifier.fit(X_train,y_train)
```

```python
y_pred=classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

score=accuracy_score(y_pred=y_pred,y_true=y_test)
cm=confusion_matrix(y_test,y_pred)
print(score)
print(cm)
print(classification_report(y_test,y_pred))
```

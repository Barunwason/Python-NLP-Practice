```python
words=['eating','eaten','eats','programs','programming','writing','writes','history','finally','finalize']
```

### Porter Stemmer

```python
from nltk.stem import PorterStemmer

stemming=PorterStemmer()

for word in words:
    print(word +"--->"+ stemming.stem(word))
```

### Regexp stemmer

```python
from nltk.stem import RegexpStemmer

reg_stemming = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemming.stem('playing')
```

```python
reg_stemming.stem('acceptable')
```

```python
reg_stemming.stem('eats')
```

### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

snow_stemming = SnowballStemmer('english')
```

```python
print("Snowball stemming --> "+snow_stemming.stem('fairly')+"\nPorter stemming -->"+stemming.stem('fairly'))
```

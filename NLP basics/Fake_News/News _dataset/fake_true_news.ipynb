```python
from nltk.stem import WordNetLemmatizer
import re
lemmatizer = WordNetLemmatizer()
corpus1=[]
for i in range(len(df)):
    review=re.sub('[^a-zA-Z5]',' ',df['title'][i])
    review= review.lower()
    review=review.split()
    review=[lemmatizer.lemmatize(word) for word in review]
    review=' '.join(review)
    corpus1.append(review)
```

```python
from nltk.stem import WordNetLemmatizer
import re
lemmatizer = WordNetLemmatizer()
corpus2=[]
for i in range(len(df)):
    review=re.sub('[^a-zA-Z5]',' ',df['text'][i])
    review= review.lower()
    review=review.split()
    review=[lemmatizer.lemmatize(word) for word in review]
    review=' '.join(review)
    corpus2.append(review)
```

```python
import numpy as np
```

```python
def avgWord2Vec(doc):
    vectors=[model.wv[word] for word in doc  if word in model.wv.index_to_key]
    if(len(vectors)==0):
        return np.zeros(model.vector_size)
    return np.mean(vectors,axis=0)
```

```python
from tqdm import tqdm
X1=[]
for i in tqdm(range(len(words1))):
    X1.append(avgWord2Vec(words1[i]))
```

```python
X2=[]
for i in tqdm(range(len(words2))):
    X2.append(avgWord2Vec(words2[i]))
```

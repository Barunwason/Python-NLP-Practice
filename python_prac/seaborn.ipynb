```python
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
tips.info()
```

```python
numerical_feature=[col for col in tips.columns if tips[col].dtype=='int64' or tips[col].dtype=='float64']
print('there are {} numerical_feature in dataset : {}'.format(len(numerical_feature),numerical_feature))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(15,5))
for i in range(len(numerical_feature)):
    plt.subplot(1,3,i+1)
    sns.kdeplot(x=tips[numerical_feature[i]],shade=True,color='red')
    plt.tight_layout()
```

```python
import matplotlib.pyplot as plt
```

```python
x=[0,1,2,4,6,8]
y=[0,2,3,4,6,7]
plt.plot(x,y,color='green',linestyle='--',marker='o',linewidth=1)
plt.grid(True)
plt.xlabel('barun')
plt.ylabel('bismeet')
plt.title("line graph")
```

```python
x=[0,1,2,4,6,8]
y1=[0,2,3,4,6,7]
y2=[0,1,2,4,6,8]
plt.figure(figsize=(9,4))
plt.subplot(1,2,1)
plt.plot(x,y1,color='green',linestyle='--',marker='o',linewidth=1)
plt.grid(True)
plt.subplot(1,2,2)
plt.plot(list(map(lambda x: (x ** 2)-3, x)),y2,color='red',linestyle='--',marker='o',linewidth=1)
plt.grid(True)
```

### Bar Graph

```python
categories=['A','B','C','D','E','F']
plt.bar(categories,x,color='red')
```

### Histogram

```python
categories=['A','B','A','D','A','F','C','C','E']

l=[1,1,2,2,2,2,3,3,4,5,5,5]
plt.figure(figsize=(9,4))
plt.hist(l,bins=5,edgecolor='black',kde=True)
```

### Scatter Chart

```python
plt.scatter(x,y,marker='x')
```

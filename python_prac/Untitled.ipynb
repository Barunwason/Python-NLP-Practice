```python
import graphviz

# Define the flowchart
flowchart = graphviz.Digraph(format='png')
flowchart.attr(rankdir='LR', size='8')

# Nodes
flowchart.node('Start', shape='circle')
flowchart.node('Frontend', 'Frontend\n(HTML, CSS, JS, Bootstrap)', shape='box')
flowchart.node('Backend', 'Backend\n(Node.js, Express)', shape='box')
flowchart.node('Database', 'MongoDB\n(Mongoose)', shape='cylinder')
flowchart.node('Mapbox', 'Mapbox API', shape='parallelogram')
flowchart.node('Cloudinary', 'Cloudinary\n(Image Uploads)', shape='parallelogram')
flowchart.node('Auth', 'Authentication\n(Passport.js)', shape='box')
flowchart.node('PropertyMgmt', 'Property Listings\n(Create/View/Delete)', shape='box')
flowchart.node('Reviews', 'Reviews\n(Add/Delete)', shape='box')
flowchart.node('MapIntegration', 'Map Integration\n(Mapbox)', shape='box')
flowchart.node('EndUser', 'End User\n(User Interface)', shape='ellipse')

# Edges
flowchart.edges([
    ('Start', 'EndUser'),
    ('EndUser', 'Frontend'),
    ('Frontend', 'Backend'),
    ('Backend', 'Database'),
    ('Backend', 'Cloudinary'),
    ('Backend', 'Mapbox'),
    ('Backend', 'Auth'),
    ('Backend', 'PropertyMgmt'),
    ('Backend', 'Reviews'),
    ('Frontend', 'MapIntegration'),
    ('MapIntegration', 'Mapbox')
])

# Render to PNG
flowchart.render('airbnb_clone_flowchart', format='png', cleanup=True)
```
